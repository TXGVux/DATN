```python
import pandas as pd
```

```python
df = pd.read_excel('Diện tích dân số.xlsx', sheet_name='Sheet1')
df.info()
df
```

```python
df.drop(columns=['Diện tích(Km2)(*)', 'Diện tích (Km2)(*)'], inplace=True)
df
```

```python
df = df[(df['Địa phương'].str.count(' ') <= 2) & (~df['Địa phương'].isin(['CẢ NƯỚC', 'Đông Nam Bộ']))]
df
```

```python
# Bước 1: Chuyển '..' thành NaN
df['Diện tích(Km2)'] = df['Diện tích(Km2)'].replace('..', pd.NA)

# Bước 1: Điền giá trị NaN bằng giá trị trước đó (ffill)    
df['Diện tích(Km2)'] = df['Diện tích(Km2)'].fillna(method='ffill')

# Bước 2: Đổi dấu phẩy sang chấm để xử lý số thập phân
df['Diện tích(Km2)'] = df['Diện tích(Km2)'].str.replace(',', '.', regex=False).astype(float)

df
```

```python
df.info()
```

```python
df['Mật độ dân số (Người/km2)'] = df['Mật độ dân số (Người/km2)'].str.replace(',', '.', regex=False).astype(float)
df['Dân số trung bình (Nghìn người)'] = df['Dân số trung bình (Nghìn người)'].str.replace(',', '.', regex=False).astype(float)
df
```

```python
df['Dân số trung bình (Nghìn người)'] = df['Dân số trung bình (Nghìn người)']*1000
df.rename(columns={'Dân số trung bình (Nghìn người)': 'Dân số trung bình'}, inplace=True)
df
```

```python
df.to_csv('dientich_danso_clean.csv', index=False, encoding='utf-8-sig')
```
