```python
import pandas as pd
import re
```

```python
df = pd.read_csv('luongcuatungnganh.csv', encoding='utf-8')
df.info()
```

```python
df
```

```python
df['KN'] = df['Muc_luong'].apply(lambda x: str(x).split(':')[0].strip() if pd.notnull(x) else None)
```

```python
def map_trinh_do(s):
    s = str(s).lower()

    if '0-2' in s:
        return 'Fresher'
    elif '3-5' in s:
        return 'Junior'
    elif 'trên 5' in s:
        return 'Senior'
    elif 'fresher/intern' in s:
        return 'Fresher'
    elif 'fresher' in s:
        return 'Fresher'
    elif 'intern' in s:
        return 'Intern'
    elif 'junior' in s:
        return 'Junior'
    elif 'senior' in s:
        return 'Senior'
    else:
        return s

# Tạo cột mới hoặc thay thế cột hiện tại
df['Trinh_do'] = df['KN'].apply(map_trinh_do)

# Xoá các dòng không xác định được trình độ
df.dropna(subset=['Trinh_do'], inplace=True)
```

```python
df['Luong_TB'] = df['Muc_luong'].apply(lambda x: str(x).split(':')[-1].strip() if pd.notnull(x) else None)
```

```python
def extract_num(s):
    # Tìm tất cả các con số, ví dụ: 15, 20, 15.5, 20,5
    numbers = re.findall(r'\d+(?:[.,]?\d*)?', s.lower())
    
    if numbers:
        numbers = [float(n.replace(',', '.')) for n in numbers]
        return sum(numbers) / len(numbers)
    
    return None
df['Lương_TB(triệu VND)'] = df['Luong_TB'].apply(extract_num)
```

```python
df.info()
```

```python
df = df[['Nganh_nghe', 'Cong_viec', 'Trinh_do', 'Lương_TB(triệu VND)']]
```

```python
df.to_csv('luongcuatungnganh_cleaned.csv', index=False, encoding='utf-8-sig')
```
